# 1. Import Configurări și Librării

```python
import pandas as pd
import numpy as np
from config import *
from pathlib import Path

print("✅ Librării încărcate cu succes.")
```

# 2. Colectare Toate Tabelele

```python
print("=" * 70)
print("COLECTARE TABELE PENTRU DISERTAȚIE")
print("=" * 70)

# Load all exported tables
tables = {}

# Table 1: Sentiment Distribution
if (DATA_OUTPUT / 'table_1_sentiment_distribution.csv').exists():
    tables['Tabel 1: Distribuție Sentiment'] = pd.read_csv(DATA_OUTPUT / 'table_1_sentiment_distribution.csv')
    print("✅ Tabel 1 încărcat")

# Table 2: Sentiment by Candidate
if (DATA_OUTPUT / 'table_2_sentiment_by_candidate.csv').exists():
    tables['Tabel 2: Sentiment pe Candidați'] = pd.read_csv(DATA_OUTPUT / 'table_2_sentiment_by_candidate.csv')
    print("✅ Tabel 2 încărcat")

# Table 3: Statistical Tests
if (DATA_OUTPUT / 'table_3_statistical_tests.csv').exists():
    tables['Tabel 3: Teste Statistice RQ1-RQ2'] = pd.read_csv(DATA_OUTPUT / 'table_3_statistical_tests.csv')
    print("✅ Tabel 3 încărcat")

# Table 4: Topics by Candidate
if (DATA_OUTPUT / 'table_4_topics_by_candidate.csv').exists():
    tables['Tabel 4: Topic-uri pe Candidați'] = pd.read_csv(DATA_OUTPUT / 'table_4_topics_by_candidate.csv')
    print("✅ Tabel 4 încărcat")

# Table 5: Spearman Correlations
if (DATA_OUTPUT / 'table_5_spearman_correlations.csv').exists():
    tables['Tabel 5: Corelații Spearman RQ3'] = pd.read_csv(DATA_OUTPUT / 'table_5_spearman_correlations.csv')
    print("✅ Tabel 5 încărcat")

# Table 6: Chi-Square Tests
if (DATA_OUTPUT / 'table_6_chi_square_tests.csv').exists():
    tables['Tabel 6: Teste Chi-Pătrat RQ3'] = pd.read_csv(DATA_OUTPUT / 'table_6_chi_square_tests.csv')
    print("✅ Tabel 6 încărcat")

print(f"\n📊 Total tabele colectate: {len(tables)}")
```

# 3. Export în Format Markdown (pentru Disertație)

```python
print("\n" + "=" * 70)
print("EXPORT ÎN FORMAT MARKDOWN")
print("=" * 70)

# Create markdown file
markdown_content = "# Rezultate Statistice - Disertație de Master\n\n"
markdown_content += "**Facultatea de Sociologie, UBB Cluj-Napoca**  \n"
markdown_content += "**Master: Analiza Datelor Complexe**  \n"
markdown_content += "**Masterand: Răzvan-Andrei Enache**  \n\n"
markdown_content += f"**Data generare:** {pd.Timestamp.now().strftime('%Y-%m-%d %H:%M')}  \n\n"
markdown_content += "---\n\n"

for name, table in tables.items():
    markdown_content += f"## {name}\n\n"
    markdown_content += table.to_markdown(index=False) + "\n\n"
    markdown_content += "---\n\n"

# Save markdown
with open(DATA_OUTPUT / 'statistical_results.md', 'w', encoding='utf-8') as f:
    f.write(markdown_content)

print(f"✅ Markdown exportat: {DATA_OUTPUT / 'statistical_results.md'}")
```

# 4. Export în Format LaTeX (Opțional)

```python
print("\n" + "=" * 70)
print("EXPORT ÎN FORMAT LATEX (OPȚIONAL)")
print("=" * 70)

latex_content = "% Rezultate Statistice - Disertație de Master\n"
latex_content += "% Generat automat\n\n"

for name, table in tables.items():
    latex_content += f"% {name}\n"
    latex_content += table.to_latex(index=False, caption=name, label=f"tab:{name.lower().replace(' ', '_')[:30]}") + "\n\n"

# Save LaTeX
with open(DATA_OUTPUT / 'statistical_results.tex', 'w', encoding='utf-8') as f:
    f.write(latex_content)

print(f"✅ LaTeX exportat: {DATA_OUTPUT / 'statistical_results.tex'}")
```
